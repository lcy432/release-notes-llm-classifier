```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()  # 自动找当前目录的 .env 文件
api_key = os.getenv("DEEPSEEK_API_KEY")

# 防御性检查
assert api_key is not None, "❌ 没读到 API key，检查 .env 文件名和内容"
print(f"✅ API key 已加载（前 8 位）: {api_key[:8]}...")
```

```python
client = OpenAI(
    api_key=api_key,
    base_url="https://api.deepseek.com"
)

response = client.chat.completions.create(
    model="deepseek-chat",
    messages=[
        {"role": "user", "content": "Reply with the word 'OK' and nothing else."}
    ],
    max_tokens=10
)

print("DeepSeek 返回：", response.choices[0].message.content)
```
